```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 1. Real-life Men's 100m World Record progression (ratified electronic/official times)
data = {
    "Year": [1912, 1921, 1930, 1936, 1956, 1960, 1968, 1983, 1988, 1991, 1994, 1996, 1999, 2005, 2008, 2009],
    "Athlete": [
        "Donald Lippincott", "Charlie Paddock", "Percy Williams", "Jesse Owens",
        "Willie Williams", "Armin Hary", "Jim Hines", "Calvin Smith",
        "Carl Lewis", "Leroy Burrell", "Leroy Burrell", "Donovan Bailey",
        "Maurice Greene", "Asafa Powell", "Usain Bolt", "Usain Bolt"
    ],
    "Time_Seconds": [10.6, 10.4, 10.3, 10.2, 10.1, 10.0, 9.95, 9.93, 9.92, 9.90, 9.85, 9.84, 9.79, 9.77, 9.69, 9.58]
}

df = pd.DataFrame(data)

# 2. Calculate total improvement and average speed (meters per second)
df["Speed_mps"] = (100 / df["Time_Seconds"]).round(2)
print("Fastest 5 World Records in History:")
print(df.tail(5)[["Year", "Athlete", "Time_Seconds", "Speed_mps"]])

# 3. Train a simple Linear Regression model to see the historical trend
X = df[["Year"]]
y = df["Time_Seconds"]
model = LinearRegression()
model.fit(X, y)

# Predict what the linear trend says for 2028 Olympics
pred_2028 = model.predict([[2028]])[0]
print(f"\nLinear Trend Projection for 2028: {pred_2028:.2f} seconds")

# 4. Plot the real data + regression trendline
plt.figure(figsize=(10, 5))
plt.scatter(df["Year"], df["Time_Seconds"], color="royalblue", label="Actual World Record")
plt.plot(df["Year"], model.predict(X), color="tomato", linestyle="--", label="Linear Regression Trend")

# Annotate Usain Bolt's 9.58s record
plt.annotate("Usain Bolt (9.58s)", xy=(2009, 9.58), xytext=(1990, 9.62),
             arrowprops=dict(arrowstyle="->", color="black"))

plt.title("Men's 100m World Record Progression (1912–2009)")
plt.xlabel("Year")
plt.ylabel("100m Time (Seconds)")
plt.legend()
plt.savefig("100m_progression.png", dpi=300, bbox_inches="tight")
plt.show()
```

Key Takeaways: Between 1912 and 2009, the Men's 100m world record dropped from 10.60s to 9.58s (an average speed of 10.44 m/s). While a linear regression model projects a 9.54s record by 2028, human sprinting speed approaches a biological limit that a straight line cannot capture.
